```python
import os
from ultralytics import YOLO

# Define the base results directory
project_dir = "killfeed/results"
base_name = "run"

# Find the next available run number
existing_runs = [d for d in os.listdir(project_dir) if d.startswith(base_name) and d[len(base_name):].isdigit()]
run_number = max([int(d[len(base_name):]) for d in existing_runs], default=0) + 1
run_name = f"{base_name}{run_number}"

# Load a model
model = YOLO("yolo11n.pt")  # Load a pretrained model

# Train the model and save results to 'killfeed/results/runX'
results = model.train(
    data="killfeed/dataset2/data.yaml",
    epochs=10,
    imgsz=448,  # Adjusted to nearest square size, killfeed images are 430x350
    patience=5,
    project=project_dir,
    name=run_name
)

print(f"Training results saved to: {os.path.join(project_dir, run_name)}")
```
