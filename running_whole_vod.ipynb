Imports

```python
from ultralytics import YOLO
import cv2
import os
```

Calling Our Models

```python
Usefullclassification_model = YOLO("runs/classify/train/weights/best.pt")
KillfeedPresent_model = YOLO("killfeed/results/run21/weights/best.pt")
```

```python
vod = "VODS/G2 vs. SEN  - VCT Americas Kickoff - Day 12 - Map 04.webm"
output_path = "whole_vod_useful.mp4"

cap = cv2.VideoCapture(vod)  # Open video

# Get video properties
fps = int(cap.get(cv2.CAP_PROP_FPS))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
print(fps,width,height)

# Define output video writer
fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # Codec for mp4 format
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))  # Video writer object

frame_skip = 10  # Process every 10th frame
frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    if frame_count % frame_skip == 0:
        # Run model inference here
            # Run classification model
        results = Usefullclassification_model(frame)
    frame_count += 1

    # Check if results contain valid probabilities
    if hasattr(results[0], "probs"):
        prediction = results[0].probs.top1  # Get predicted class (1 = gameplay, 0 = not gameplay)
        
        if prediction == 1:
            out.write(frame)  # Write only frames classified as gameplay

# Release resources
cap.release()
out.release()
cv2.destroyAllWindows()

print(f"Processed video saved as {output_path}")
```
